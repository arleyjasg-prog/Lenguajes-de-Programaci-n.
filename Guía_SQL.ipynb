# SQL

Una base de datos puede definirse como un conjunto de información organizada
mediante estructuras de datos en la que se pueden ejecutar diferentes operaciones
para realizar consultas o visualizar información (Date, 2001).

Para la gestión de bases de datos existen varios lenguajes de programación y, a su
vez, diversos programas (software), los cuales permiten realizar diferentes operaciones
sobre la información contenida en tales bases de datos de manera estructurada y
eficiente. El lenguaje más utilizado para esta gestión es SQL (lenguaje de consultas
estructurado, por sus siglas en inglés structured query language) (Joshi, Gundale y
Jagdale, 2012); para trabajar tal lenguaje se propone el programa MySQL, de carácter
gratuito.
A lo largo de esta unidad, se presenta la información sobre los componentes de una
base de datos, los tipos de relaciones que se pueden crear y las operaciones básicas
que se realizan en la gestión de bases de datos.



# Modelo Relacional

Al momento de generar una base de datos, una de las decisiones más importantes
es determinar el tipo de modelo de datos a utilizar, el cual determina la estructura que
va a adoptar la información dentro de la base de datos, por lo que cada dato dentro de
esta posee un conjunto de propiedades dadas principalmente por el modelo de datos
utilizado (Marqués, 2009).
A través de la historia se han propuesto diversos modelos de datos, sin embargo,
el relacional, propuesto por Edgar Codd en 1970, ha sido el más utilizado. Como
su nombre lo indica, este modelo está basado en relaciones, las cuales pueden ser
entendidas como tablas de datos organizadas en columnas y filas; es decir que, bajo
un modelo relacional, una base de datos es un conjunto de tablas de datos (Quiroz,
2003).
Dentro del modelo cada una de las tablas es independiente, y dentro de su
elaboración bidimensional se debe tener en cuenta que se denominan atributos a las
columnas y registros a las filas. En la figura 1 se puede observar la estructura de una
base de datos bajo un modelo relacional por medio de un ejemplo.



# Relaciones base

Las relaciones base o relaciones reales son aquellas que poseen un nombre y que están
definidas como parte estructural dentro de la base de datos (Marqués, 2009). En el ejemplo
de la figura 1 se pueden notar dos relaciones base (clientes y ciudades), ya que ambas están
definidas e ingresadas directamente dentro de la base de datos.
Al momento de generar relaciones base, se deben tener en cuenta los siguientes
elementos o características:
1. **Nombres de atributos:** estos nombres deben cumplir dos condiciones: no pueden
contener espacios y todos deben ser diferentes. Además, por facilidad de manejo, se
recomienda asignar nombres representativos; por ejemplo, teniendo en cuenta la figura
1, se puede asignar al atributo Razón social los nombres RazSocial o Razon_Social.
2. **Claves:** dentro de las tablas no pueden existir registros duplicados, por lo que se
deben especificar los atributos que hacen únicas cada una de las filas de la tabla, denominados claves primarias. Por ejemplo, en la figura 1, dentro de la relación
CLIENTE se puede asignar el atributo ID_Cliente por clave primaria. Por otro lado,
se deben identificar los atributos que puedan ser compartidos entre una tabla
y otra, con el objetivo de crear vínculos entre ellas, estos se denominan claves
foráneas. Por ejemplo, en la figura 1, se puede generar el atributo Codigo_Ciudad
como clave foránea de la relación CIUDADES dentro de la relación CLIENTE.

3. **Tipos de datos:** para cada uno de los atributos de las tablas debe ingresarse el
tipo de datos que estas pueden contener (texto, números, fechas o algún formato
adicional). En la tabla 1 se muestran los principales tipos de datos y cómo se
deben asociar a un atributo.

4. **Esquema:** este sirve para denotar la estructura de la base de datos de manera
escrita, especificando los atributos que se utilizan dentro de cada relación. Del
ejemplo mostrado en la figura 1 se puede elaborar el siguiente esquema, donde
se subrayan las claves primarias y se encuentran en negrita las claves foráneas.
CLIENTES (ID_Cliente, Razon_Social, Direccion, CodCiudad, Telefono)
CIUDADES (CodCiudad, Nombre_Ciudad, CodProd)

# Relaciones derivadas

Partiendo de las relaciones o tablas generadas para una base de datos se
pueden generar nuevas relaciones como resultado de la aplicación de un número
de operaciones; estas tablas son llamadas relaciones derivadas (Ramírez y Laguna,
2014). De esta forma, las relaciones derivadas se pueden entender como las tablas que
resultan al realizar una consulta sobre una base de datos.
Las consultas dentro de una base de datos se realizan por medio de operaciones
lógicas efectuadas sobre las relaciones, esto se conoce como álgebra relacional
(Quiroz, 2003).
En los siguientes apartados se muestran cada una de las operaciones más utilizadas
para crear relaciones derivadas, utilizando el software MySQL y mediante ejemplos
prácticos.

# Joins de tablas

Hasta aquí se ha explicado la generación de relaciones derivadas como resultado
de una consulta efectuada sobre una tabla. Sin embargo, dichas relaciones pueden
obtenerse de igual manera por medio de operaciones sobre las tablas pertenecientes a una base de datos.

Los joins de tablas hacen parte de las operaciones más utilizadas para crear
relaciones derivadas dentro de bases de datos. A continuación se explican tres joins que se pueden realizar sobre tablas.

## INNER JOIN

Esta operación se utiliza para generar relaciones derivadas entre dos tablas que
comparten algún atributo, es decir, a partir de una relación base, generar una tabla únicamente con los registros que contienen un valor dentro de un atributo que es compartido con otra relación base.

Para ejecutar esta operación, inicialmente se debe escribir la instrucción SELECT, la cual debe estar acompañada de las columnas de la tabla 1 que se desean mostrar en el resultado; posteriormente se debe indicar el nombre de la tabla 1 por medio de la instrucción FROM; luego se debe utilizar la instrucción INNER JOIN, referenciando, además, el nombre de la tabla 2; y finalmente, se debe utilizar la instrucción ON para denotar los atributos que coinciden en cada tabla. La sintaxis de esta función es la siguiente:

SELECT atributo1, atributo2…
FROM [Nombre tabla 1]
INNER JOIN [Nombre tabla 2]
ON atributo tabla 1 = atributo tabla 2;

```python
consulta = """
SELECT
       v.Fecha,
       v.Vendedor,
       p.Nombre_producto,
       p.Precio,
       v.Cantidad
FROM ventas AS v
INNER JOIN Productos AS p
    ON v.Cod_Prod = p.Cod_Prod;
"""

df_join = pd.read_sql_query(consulta, conn)
df_join
```

**Crear un nuevo atributo a partir de atributos de diferentes tablas**
Total_Venta=Cantidad*Precio

```python
consulta = """
SELECT
       v.Fecha,
       v.Vendedor,
       p.Nombre_producto,
       p.Precio,
       v.Cantidad,
       v.Cantidad * p.Precio AS Total_Venta
FROM ventas AS v
INNER JOIN Productos AS p
    ON v.Cod_Prod = p.Cod_Prod;
"""

df_join = pd.read_sql_query(consulta, conn)
df_join
```

## LEFT JOIN



```python
# A) En VENTAS: cambiamos una venta a un código inexistente en Productos
cur.execute("""
UPDATE ventas
SET Cod_Prod = 'P-999'        -- no existe en Productos
WHERE Cod_venta = 1005;       -- antes: P-004
""")

# B1) En PRODUCTOS: renombrar P-004 -> P-044 (rompe match con ventas que aún dicen P-004)
cur.execute("""
UPDATE Productos
SET Cod_Prod = 'P-044'
WHERE Cod_Prod = 'P-004';
""")

# B2) En PRODUCTOS: renombrar P-003 -> P-033 (deja a Productos con un código sin ventas;
# y las ventas que tenían P-003 ahora quedan "huérfanas")
cur.execute("""
UPDATE Productos
SET Cod_Prod = 'P-033'
WHERE Cod_Prod = 'P-003';
""")

conn.commit()
```

Ver efecto de LEFT JOIN (todas las ventas)

```python

consulta_left = """
SELECT v.Cod_venta, v.Fecha, v.Vendedor, v.Cod_Prod,
       p.Nombre_producto, p.Precio
FROM ventas AS v
LEFT JOIN Productos AS p
  ON v.Cod_Prod = p.Cod_Prod
ORDER BY v.Cod_venta;
"""
df_left = pd.read_sql_query(consulta_left, conn)
df_left
```

Simular RIGHT JOIN (todas los productos) invirtiendo el LEFT JOIN

```python
consulta_right_sim = """
SELECT p.Cod_Prod, p.Nombre_producto, p.Precio,
       v.Cod_venta, v.Fecha, v.Vendedor, v.Cantidad
FROM Productos AS p
LEFT JOIN ventas AS v
  ON v.Cod_Prod = p.Cod_Prod
ORDER BY p.Cod_Prod;
"""
df_right = pd.read_sql_query(consulta_right_sim, conn)
df_right
```

## Agrupamiento con GROUP BY

La cláusula `GROUP BY` en SQL se utiliza para agrupar filas que tienen los mismos valores en una o más columnas. Esto se hace a menudo en conjunto con funciones de agregación (como `COUNT()`, `SUM()`, `AVG()`, `MIN()`, `MAX()`) para realizar cálculos en cada grupo de filas.

Imagina que tienes una tabla de ventas y quieres saber la cantidad total vendida por cada vendedor. No te interesa ver cada venta individual, sino el resumen por vendedor. Aquí es donde `GROUP BY` es útil.

La sintaxis básica es:

```python
consulta_group_by = """
SELECT Vendedor, SUM(Cantidad) AS Total_Cantidad_Vendida
FROM ventas
GROUP BY Vendedor
ORDER BY Total_Cantidad_Vendida DESC;
"""

df_group_by = pd.read_sql_query(consulta_group_by, conn)
df_group_by
```

```python
consulta_cantidad_por_producto = """
SELECT Cod_Prod, SUM(Cantidad) AS Total_Cantidad_Vendida
FROM ventas
GROUP BY Cod_Prod
ORDER BY Total_Cantidad_Vendida DESC;
"""

df_cantidad_por_producto = pd.read_sql_query(consulta_cantidad_por_producto, conn)
df_cantidad_por_producto
```

```python
consulta_ventas_por_vendedor = """
SELECT
    v.Vendedor,
    SUM(v.Cantidad * p.Precio) AS Total_Ventas_Dinero
FROM ventas AS v
INNER JOIN Productos AS p
    ON v.Cod_Prod = p.Cod_Prod
GROUP BY v.Vendedor
ORDER BY Total_Ventas_Dinero DESC;
"""

df_ventas_por_vendedor = pd.read_sql_query(consulta_ventas_por_vendedor, conn)
df_ventas_por_vendedor
```

```python
# Suponiendo que ya tienes un archivo de Excel llamado 'mi_archivo.xlsx'
# en el directorio /content/

# Paso 1: Cargar el archivo de Excel en un DataFrame de pandas
try:
    # *** MODIFICA LA RUTA DEL ARCHIVO AQUI ***
    # Ejemplo: '/content/drive/My Drive/mi_archivo.xlsx'
    # Si está en una subcarpeta: '/content/drive/My Drive/Carpeta/mi_archivo.xlsx'
    df_excel = pd.read_excel('/content/drive/My Drive/excel_SQL.xlsx')
    print("Archivo de Excel cargado exitosamente.")
    display(df_excel.head())

    # Paso 2: Guardar el DataFrame en la base de datos SQLite
    # 'replace' eliminará la tabla si ya existe y creará una nueva
    # 'index=False' evita que pandas escriba el índice del DataFrame como una columna en SQL
    df_excel.to_sql('nueva_tabla_desde_excel', conn, if_exists='replace', index=False)

    print("\nDataFrame guardado como tabla SQL 'nueva_tabla_desde_excel'.")

    # Verificar que la tabla fue creada leyendo de ella
    df_sql_verificacion = pd.read_sql_query("SELECT * FROM nueva_tabla_desde_excel LIMIT 5;", conn)
    print("\nDatos de la nueva tabla SQL:")
    display(df_sql_verificacion)

except FileNotFoundError:
    print("Error: Asegúrate de que la ruta del archivo sea correcta y que el archivo exista en tu Google Drive.")
except Exception as e:
    print(f"Ocurrió un error: {e}")
```
